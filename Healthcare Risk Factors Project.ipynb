### __📌Outliers__

Outliers are extreme values that are very different from the rest of the data.

```python
#Function to Detect Outliers for All Numeric Columns

def detect_outliers(df):
    outlier_summary = {}

    for col in df.select_dtypes(include=['int64','float64']).columns:
        
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        
        outlier_summary[col] = len(outliers)
    
    return outlier_summary
   
```

```python
outliers = detect_outliers(df)
print(outliers)
```

__Cap the Outliers (Winsorization)__



```python
Q1 = df['Glucose'].quantile(0.25)
Q3 = df['Glucose'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['Glucose'] = df['Glucose'].clip(lower, upper)  # Clip the Outliers Q1 = 95 ,Q3 = 135 ,IQR = 40

#Values below lower bound → replaced with lower
#Values above upper bound → replaced with upper
```

```python
import seaborn as sns
sns.boxplot(x=df['Glucose'])
```

#### 📌__Insights__




Outliers in the glucose variable were identified using the IQR method and visualized using a boxplot. Since 
high glucose values may represent real diabetic cases rather than data errors, the outliers were capped 
using the IQR upper and lower bounds instead of removing them to preserve important clinical information.

### __📌Outliers__

Outliers are extreme values that are very different from the rest of the data.

```python
#Function to Detect Outliers for All Numeric Columns
# The function scans all numeric columns and returns how many outliers exist in each column using the IQR method.

def detect_outliers(df):
    outlier_summary = {}  # dictionary used to store results.

    for col in df.select_dtypes(include=['int64','float64']).columns:  # selects numeric columns (int64, float64) from the dataset.
        
        Q1 = df[col].quantile(0.25)   #For each column, it calculates Q1 (25th percentile) and Q3 (75th percentile).
        Q3 = df[col].quantile(0.75)
        
        IQR = Q3 - Q1   # (Interquartile Range) method.--> measure data spread.

        # It defines lower and upper bounds to detect extreme values.
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR  # Any value below the lower bound or above the upper bound is considered an outlier.
        
        
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]  # it counts the number of outliers in each column and returns the summary.

        #stores the number of outliers for each column in the dictionary.--> outlier_summary[col]
        
        outlier_summary[col] = len(outliers) # counts how many outlier rows were found.

    
    return outlier_summary
   
```

```python
outliers = detect_outliers(df)
print(outliers)
```

__Cap the Outliers (Winsorization)__



```python
Q1 = df['Glucose'].quantile(0.25)
Q3 = df['Glucose'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['Glucose'] = df['Glucose'].clip(lower, upper)  # Clip the Outliers Q1 = 95 ,Q3 = 135 ,IQR = 40

#Values below lower bound → replaced with lower
#Values above upper bound → replaced with upper
```

```python
import seaborn as sns
sns.boxplot(x=df['Glucose'])
```

📌__Insights :__


__Outliers in the glucose__ variable were identified using the __IQR method and visualized using a boxplot__.

Since __high glucose values may represent real diabetic cases rather than data errors__, the __outliers were capped 
using the IQR upper and lower bounds instead of removing them to preserve important clinical information.__
